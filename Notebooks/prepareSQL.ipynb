```python
import pandas as pd

# Original cleaned dataset (before one-hot encoding)
cleaned_df = pd.read_csv("../Data/Telco_Customer_Churn_processed.csv")

encoded_df = pd.read_csv("../Data/model_results.csv")
```

```python
print(cleaned_df.columns.tolist())
```

```python
sql_df = cleaned_df.copy()

sql_df["churn_probability"] = encoded_df["churn_probability"]
sql_df["risk_segment"] = encoded_df["risk_segment"]

sql_df["annual_value"] = sql_df["MonthlyCharges"] * 12
sql_df["expected_revenue_at_risk"] = (
    sql_df["annual_value"] * sql_df["churn_probability"]
)
```

```python
sql_df.to_csv("../Data/model_results_sql.csv", index=False)
```
